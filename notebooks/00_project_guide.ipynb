# Hướng dẫn sử dụng project Bài tập 2 — Thị giác máy tính

Project chia thành 5 thuật toán và 5 notebook tương ứng. Mỗi file `.py` chỉ định nghĩa thuật toán; notebook giải thích, gọi hàm và kiểm thử.

## Cây thư mục

```text
Computer_Vision_BT2/
├── src/cv_bt2/corner/          # Harris, Harris-Laplace, ANMS, metrics
├── src/cv_bt2/segmentation/    # Region Growing, SRG, metrics
├── src/cv_bt2/data/            # ảnh tổng hợp tái lập
├── notebooks/                  # 5 notebook thuật toán + notebook tổng hợp
├── scripts/                    # chạy thí nghiệm từ terminal
├── tests/                      # unit tests
├── data/images/                # ảnh thật người dùng có thể thêm
└── outputs/                    # hình và bảng được sinh ra
```

## Cài đặt và chạy

Mở terminal tại thư mục `Computer_Vision_BT2`:

```bash
python -m venv .venv
# Windows: .venv\Scripts\activate
# macOS/Linux: source .venv/bin/activate
pip install -r requirements.txt
pip install -e .
pytest -q
jupyter lab
```

Mở lần lượt `01` đến `05` để học từng thuật toán, sau đó chạy `06_integrated_comparison.ipynb`. Nên đặt working directory ở thư mục gốc project.

## File thuật toán và notebook tương ứng

| Thuật toán | File định nghĩa | Notebook giải thích/test |
|---|---|---|
| Harris cổ điển | `corner/harris_classic.py` | `01_harris_classic.ipynb` |
| Harris–Laplace | `corner/harris_laplace.py` | `02_harris_laplace.ipynb` |
| ANMS | `corner/anms.py` | `03_anms.ipynb` |
| Region Growing | `segmentation/region_growing.py` | `04_region_growing.ipynb` |
| SRG | `segmentation/srg.py` | `05_seeded_region_growing.ipynb` |

`corner/metrics.py`, `segmentation/metrics.py`, `data/synthetic.py` và `visualization.py` là các module hỗ trợ, không thay thế thuật toán.
